<h3>Prétraitement des données</h3>

```python
import pandas as pd 


df = pd.read_csv("./WELFake_Dataset.csv")
```

```python
print(df.shape)
```

```python
print(df.head())
```

```python
print(df.info())
```

Supprimer les colonnes inutiles

```python
df.drop(columns=['Unnamed: 0'],inplace=True)
print(df.info())
```

 Supprimer les doublons 

```python

print("Number of rows with duplicates: ",df.shape[0])
df.drop_duplicates(inplace=True)
print("Number of rows without duplicates: ",df.shape[0])
```

Vérifier et supprimer les valeurs manquantes

```python
df.isna().sum()

df.dropna(subset=['title', 'text'],inplace=True)

df.isna().sum()
```

 Séparation des variables

```python


cible = "label"
y = df[cible]

X = df.drop(cible,axis=1)

print(f"cible ou target (y) est : {cible}")
print(f"Les variables features (X) contiennent {X.shape[1]} colonnes.")

```

Séparation du jeu de données

```python
from sklearn.model_selection import train_test_split

X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, 
    test_size=0.15, 
    random_state=42, 
    stratify=y
)


X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, 
    test_size=(0.15 / (1.0 - 0.15)), 
    random_state=42, 
    stratify=y_train_val
)

print(f"Train shape: {X_train.shape}") 
print(f"Val shape:   {X_val.shape}")    
print(f"Test shape:  {X_test.shape}")  


train = pd.DataFrame(X_train).assign(label=y_train)
val = pd.DataFrame(X_val).assign(label=y_val)
test = pd.DataFrame(X_test).assign(label=y_test)

train.to_csv("./data/train.csv", index=False)
val.to_csv("./data/val.csv", index=False)
test.to_csv("./data/test.csv", index=False)
```
